```python
#Drop customer ID becuse it is only an identifier
df=df.drop("CUST_ID",axis=1)
```

```python
#check missing values
df.isnull().sum()
```

```python
#fill missing values with the median
df["CREDIT_LIMIT"] = df["CREDIT_LIMIT"].fillna(
    df["CREDIT_LIMIT"].median()
)
df["MINIMUM_PAYMENTS"] = df["MINIMUM_PAYMENTS"].fillna(
    df["MINIMUM_PAYMENTS"].median()
)
```

```python
#verify missing values
df.isnull().sum()
```

Feature scaling using StandardScaler

```python
from sklearn.preprocessing import StandardScaler
```

```python
scaler=StandardScaler()
X_scaled=scaler.fit_transform(df)
```

covert scaled data back to DataFrame

```python
X_scaled_df=pd.DataFrame(X_scaled,columns=df.columns)
X_scaled_df.head()
```

check the scaled data

```python
X_scaled_df.describe()
```

k_means clustering and elbow method

```python
from sklearn.cluster import KMeans
```

```python
wcss=[]
for k in range(2,11):
  kmeans=KMeans(n_clusters=k,random_state=42,n_init=10)
  kmeans.fit(X_scaled)
  wcss.append(kmeans.inertia_)
```

```python
wcss
```

Elbow method for optimal number of clusters

```python
plt.figure(figsize=(10,6))
plt.plot(range(2,11),wcss,marker="o")
plt.xlabel("Number of Clusters (K)")
plt.ylabel("WCSS/Inertia")
plt.title("Elbow Method for Optimal Number of Clusters")
plt.xticks(range(2,11))
plt.grid(True)
plt.show()
```

Silhouette Score Analysis

```python
from sklearn.metrics import silhouette_score
```

```python
silhouette_scores = []
for k in range(2, 11):
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
    cluster_labels=kmeans.fit_predict(X_scaled)
    score=silhouette_score(X_scaled,cluster_labels)
    silhouette_scores.append(score)

```

```python
silhouette_scores
```

silhouette graph

```python
plt.figure(figsize=(10,6))
plt.plot(range(2,11),silhouette_scores,marker="o")
plt.xlabel("Number of Clusters (K)")
plt.ylabel("Silhouette Score")
plt.title("Silhouette Score for Different Numbers of Clusters")
plt.xticks(range(2,11))
plt.grid(True)
plt.show()
```

Final K-means Clustring

```python
best_k=range(2,11)[silhouette_scores.index(max(silhouette_scores))]
print("Best K based on Silhouette Score:",best_k)
print("Highest Silhouette Score:",max(silhouette_scores))
```

```python
#set the optmil number of clusters
best_k=3
#create and train the final k-means model
final_kmeans=KMeans(n_clusters=best_k,random_state=42,n_init=10)
#Assign cluster labels to customers
df["Cluster"]=final_kmeans.fit_predict(X_scaled)
df.head()
```

```python
df["Cluster"].value_counts().sort_index()
```

Cluster wise feature analysis

```python
#calculate the mean of each feature for every cluster
cluster_means=df.groupby("Cluster").mean()
cluster_means
```

cluster -wise feature means heatmap

```python
plt.figure(figsize=(16,8))
sns.heatmap(cluster_means,annot=True,fmt=".2f",cmap="coolwarm",center=0,linewidth=0)
plt.title("Average Feature Values by Customer Cluster")
plt.ylabel("Features")
plt.ylabel("Customer Clusters")
plt.tight_layout()
plt.show()
```

```python
#disply cluster wise averge feature values
cluster_means.round(2)
```

Hierarchical Clustering

```python
#select 300 random samples from the scaled dataset
X_sample=pd.DataFrame(X_scaled).sample(n=300,random_state=42)
```

```python
X_sample.shape
```

Dendrogram Visualization

```python
from scipy.cluster.hierarchy import dendrogram,linkage
```

```python
#perform hierarchical clustring
linked=linkage(X_sample,method="ward")
#create dendrogram
plt.figure(figsize=(15,7))
dendrogram(linked)
plt.title("Hierarchical Clustering Dendrogram")
plt.xlabel("Customers Samples")
plt.ylabel("Euclidean Distances")
```

```python
plt.figure(figsize=(15,7))
dendrogram(linked)
#add horizontal threshold line
threshold=25
plt.axhline(y=threshold,linestyle="--",color="red",label=f"Threshold={threshold}")

plt.title("Hierarchical Clustering Dendrogram")
plt.xlabel("Customers Samples")
plt.ylabel("Euclidean Distances")
plt.legend()
plt.tight_layout()
plt.show()
```

Agglomerative Clustring

```python
from sklearn.cluster import AgglomerativeClustering
```

```python
#create agglimerative clustring model
agg_model=AgglomerativeClustering(n_clusters=3,linkage="ward")
#fit the model and generate cluster labels
hierarchical_labels=agg_model.fit_predict(X_scaled)
#display cluster distribution
pd.Series(hierarchical_labels).value_counts().sort_index()
```

K-Means vs Hierarchical Clustring Comparison

```python
from sklearn.cluster import KMeans
from sklearn.metrics import adjusted_rand_score
```

```python
from sklearn.cluster import AgglomerativeClustering

# k-means on the same 300row sample
kmeans_sample=KMeans(n_clusters=3,random_state=42,n_init=10)
kmeans_labels=kmeans_sample.fit_predict(X_sample)

# Perform hierarchical clustering on the same 300-row sample
agg_model_sample = AgglomerativeClustering(n_clusters=3, linkage="ward")
hierarchical_labels_sample = agg_model_sample.fit_predict(X_sample)

#calculate ARI using labels from the same sample
ari_score=adjusted_rand_score(hierarchical_labels_sample,kmeans_labels)
print("Adjusted Rand Index (ARI):",ari_score)
```
